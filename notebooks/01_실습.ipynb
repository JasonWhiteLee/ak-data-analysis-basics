## 1-2. 파이썬 — 딱 세 가지만

문법을 다 배우지 않습니다. 오늘 필요한 건 이 셋뿐입니다.

### ① 변수 — 값에 이름을 붙인다
엑셀에서 셀에 이름을 정의하는 것과 같습니다.

````python
매출 = 1500
원가 = 900

이익 = 매출 - 원가
이익
````

### ② 값에는 종류가 있다
계산이 안 되는 사고의 90%가 여기서 납니다.

- `1500` → **숫자**
- `'1500'` → **글자** (따옴표가 붙으면 숫자가 아닙니다)

아래를 실행해 보세요. 무슨 일이 일어나는지 보세요.

````python
'1500' + '900'
````

`2400` 이 아니라 `'1500900'` 이 나옵니다.
파이썬은 글자끼리 더하라고 하면 **이어 붙입니다.**

> 엑셀에서 숫자가 왼쪽 정렬되고 SUM이 0으로 나오던 그 상황과 같은 문제입니다.
> **오늘 오후에 이 문제를 실제 데이터에서 만나게 됩니다.**

### ③ 함수 — 무언가를 시키는 것
엑셀의 `SUM(...)` 과 똑같습니다. 이름 뒤에 괄호를 붙이면 실행됩니다.

````python
len('데이터분석')
````

그리고 **점(`.`)** 이 붙는 형태도 자주 나옵니다.
`무엇.무엇을해라()` — "이 값에게 이 일을 시켜라" 라는 뜻입니다.

````python
'  코드잇  '.strip()
````

## 1-4. 표를 담는 그릇 — DataFrame

pandas 에서 표는 **DataFrame** 이라고 부릅니다.
엑셀 시트 하나라고 생각하면 정확합니다.

직접 하나 만들어 보겠습니다.

````python
df = pd.DataFrame({
    '지점': ['서울', '부산', '대구'],
    '매출': [1500, 900, 700],
})

df
````

- **행(row)** — 가로줄. 여기서는 지점 하나가 한 행
- **열(column)** — 세로줄. `지점`, `매출`
- 맨 왼쪽 `0 1 2` 는 **인덱스**. 엑셀의 행 번호와 같습니다 (0부터 셉니다)

열 하나만 꺼내려면 대괄호를 씁니다.

````python
df['매출']
````

````python
df['매출'].sum()
````

## 1-5. 진짜 데이터 불러오기

이제 실제 업무 데이터를 가져옵니다.
어느 문구·가구 유통사의 **4년치 주문 내역**입니다.

파일을 내려받을 필요 없이, 주소만 알려주면 pandas 가 알아서 읽어 옵니다.

````python
BASE = 'https://raw.githubusercontent.com/JasonWhiteLee/ak-data-analysis-basics/main/'

orders = pd.read_csv(BASE + 'superstore_orders.csv', parse_dates=['Order Date', 'Ship Date'])

orders.head()
````

`head()` 는 **맨 위 5줄만** 보여 줍니다.
1만 줄짜리 표를 통째로 화면에 띄우면 아무것도 안 보이니까요.

> `parse_dates=[...]` 는 "이 열들은 날짜로 읽어라"는 뜻입니다.
> 이걸 빼면 날짜가 그냥 **글자**가 되어 버립니다. (1-2 의 ② 를 기억하세요)

## 1-6. 첫인상 — 네 가지만 확인한다

데이터를 열면 **항상 이 순서로** 봅니다.

### ① 얼마나 큰가

````python
orders.____
````

### ② 어떤 열이 있는가

````python
orders.____
````

### ③ 각 열이 무슨 타입인가
`object`(또는 `str`) 는 글자, `int64`·`float64` 는 숫자, `datetime64` 는 날짜입니다.

````python
orders.____
````

### ④ 빠진 값은 없는가
`info()` 는 위 셋을 한 번에 보여 주고, **비어 있지 않은 개수**까지 알려 줍니다.

````python
orders.info()
````

### 🔍 잠깐 — 뭔가 이상한 열이 있습니다

전체 행 수와 `Non-Null Count` 를 비교해 보세요.
**숫자가 다른 열이 두 개** 있습니다. 무엇인가요?

아래 셀로 열별 빈칸 개수를 셀 수 있습니다.

````python
orders.isna().sum()
````

> **이 빈칸들이 왜 비었는지는 2교시에 다룹니다.**
> 지금은 "빈칸이 있다"는 사실만 기억해 두세요.

## 1-8. 실습 — 이 데이터는 누구인가
