# 7-4. 3단계 · 정제 (그리고 무엇을 뺐는지 기록하기)

2교시와 4교시에서 한 것을 이어 붙입니다.

1. **중복 행** 제거
2. **매출 0 / 수량 음수** 제외 (명백한 오류)
3. **자릿수 오류** 보정 (Tables 104,835건)

빈칸 하나가 있습니다. 중복 제거하는 함수 이름입니다.

```python
n_raw = len(orders)

# (1) 중복값(모든 열이 동일한 행) 제거
clean = orders.drop_duplicates()
n_dedup = len(clean)

# (2) 매출 0 이하, 수량 0 이하인 행 제외
clean = clean[(clean['Sales'] > 0) & (clean['Quantity'] > 0)].copy()
n_clean = len(clean)

# (3) 자릿수가 100배로 잘못 들어간 것으로 보이는 건 보정
n_digit_error = (clean['Sales'] > 50000).sum()
clean.loc[clean['Sales'] > 50000, 'Sales'] = clean['Sales'] / 100

print('원본            : {:>6} 행'.format(n_raw))
print('중복 제거 후    : {:>6} 행  (-{})'.format(n_dedup, n_raw - n_dedup))
print('오류 제거 후    : {:>6} 행  (-{})'.format(n_clean, n_dedup - n_clean))
print('자릿수 보정     : {:>6} 건'.format(n_digit_error))
```

## 정제 기록 — 보고서에 그대로 옮길 내용

| 처리 | 건수 | 이유 |
|---|---|---|
| 중복값 제거 | **45건** | 모든 열이 동일한 행이라 같은 건이 두 번 들어간 것으로 판단 |
| 매출 0 또는 수량 0 이하 제외 | 24건 | 사은품인지 입력 누락인지 알 수 없어 제외 |
| 자릿수 오류 보정 | 1건 | Tables 104,835 → 1,048.35 (이익이 -69.9라 100배 오입력으로 판단) |

**남은 행: 10,170행 (원본 10,239행의 99.3%)**

> 이 표를 왜 만드나 하면, **나중에 다른 사람이 같은 데이터로 다른 숫자를 냈을 때**
> 어디서 갈렸는지 찾기 위해서입니다.
> 뺀 것은 다 합쳐 69건, 전체의 0.7%라 결론이 바뀔 정도는 아닙니다. 그래도 적습니다.
> 적는 데 30초 걸리고, 안 적으면 나중에 반나절을 씁니다.

**"뺐다"고 적는 것이 "안 뺐다"보다 신뢰를 잃는 일이 아닙니다.**
숫자가 어떻게 만들어졌는지 밝힌 보고서가 더 믿을 만한 보고서입니다.

> ### 여기서부터 3교시와 숫자가 달라집니다

# 7-5. 4단계 · merge 로 반품 정보 붙이기

3교시에서 한 `merge` 입니다.
`returns` 파일에는 **반품된 주문번호만** 들어 있습니다.

```python
print(returns.shape)
returns.head(3)
```

296건뿐입니다. **반품 안 된 주문은 아예 이 파일에 없습니다.**

그래서 `how='left'` 로 붙입니다. 왼쪽(정리본)은 다 남기고,
오른쪽에 짝이 없는 행은 결측치가 됩니다. 그 결측치를 `'No'` 로 채웁니다.

```python
clean = clean.merge(returns, on='Order ID', how='left')
clean['Returned'] = clean['Returned'].fillna('No')

print('merge 후 행 수 :', len(clean))      # 늘어나지 않아야 정상입니다
print()
print(clean['Returned'].value_counts())
```

**행 수가 10,170에서 늘지 않았습니다.** 이걸 꼭 확인하세요.

3교시에서 봤듯이, 오른쪽 표에 같은 키가 여러 개 있으면 merge 후 행이 불어납니다.
행 수를 확인하지 않으면 매출 합계가 부풀어 오른 것도 모르고 보고하게 됩니다.

# 7-6. 5단계 · 집계 — 품목별로 묶어 보기

드디어 질문에 답할 차례입니다. **품목별 매출·이익·이익률**을 냅니다.

빈칸이 두 개 있습니다. "묶는다"는 뜻의 함수와, "정렬한다"는 뜻의 함수입니다.

```python
subcat = clean.groupby('Sub-Category').agg(
    sales=('Sales', 'sum'),
    profit=('Profit', 'sum'),
    n_orders=('Sales', 'size'),
)
subcat['margin'] = (subcat['profit'] / subcat['sales'] * 100).round(1)

# 이익률이 낮은 순으로 정렬합니다
subcat.sort_values('margin').round(0)
```

## 맨 위 세 줄이 적자입니다

|---|---|---|---|
| **Tables** | 208,020 | **-17,753** | **-8.5%** |
| **Bookcases** | 115,361 | **-3,632** | **-3.1%** |
| **Supplies** | 46,725 | **-1,171** | **-2.5%** |

그리고 맨 아래를 보면 `Labels` 44%, `Envelopes` 42%, `Paper` 40%, `Copiers` 37% 로
이익률이 아주 높은 품목들이 있습니다.

**Tables 는 매출로 보면 4위권인데 이익은 전체에서 가장 많이 까먹고 있습니다.**
매출만 보고 "테이블이 잘 팔린다"고 했으면 정반대 결론을 냈을 겁니다.

## 큰 그림 — Category 3개로도 봅시다

## 6단계 · 왜 적자인가 — 할인과 반품을 붙여 봅니다

"Tables 가 적자다"까지는 알았습니다. 여기서 멈추면 할 수 있는 일이 없습니다.
**원인 후보 두 가지**를 데이터로 확인해 봅니다. 할인율과 반품률입니다.

```python
cause = clean.groupby('Sub-Category').agg(
    avg_discount=('Discount', 'mean'),
    return_incl_rate=('Returned', lambda s: (s == 'Yes').mean()),
)
cause['avg_discount'] = (cause['avg_discount'] * 100).round(1)
cause['return_incl_rate'] = (cause['return_incl_rate'] * 100).round(1)
cause['margin']  = subcat['margin']          # 앞에서 만든 표에서 가져옵니다

cause = cause[['margin', 'avg_discount', 'return_incl_rate']]
print('전체 평균 할인율: {:.1f}%'.format(clean['Discount'].mean() * 100))
cause.sort_values('margin').head(6)
```

> ### 열 이름을 '반품률'이 아니라 '반품포함률'로 붙인 이유
> 1교시에서 확인했듯이 **한 줄은 주문이 아니라 품목**입니다.
> 그리고 한 주문이 반품되면 그 주문에 담긴 품목이 **전부** 반품 처리됩니다.
>
> 그래서 위 숫자는 "이 품목이 들어 있던 줄 중 반품된 주문에 속한 비율"입니다.
> 품목별로 보려면 이 방법밖에 없지만, **주문 단위 반품률과는 값이 다릅니다.**
> 이름을 다르게 붙여 두지 않으면 나중에 두 숫자를 섞어 쓰게 됩니다.

전체 평균 할인율은 **15.5%** 입니다.
**적자 품목들은 할인율이 그보다 훨씬 높습니다.** Tables 25.8%, Machines 30.6%.

반품포함률은 품목별로 6~11% 사이라 적자 품목만 유난히 높지는 않습니다.

## 8단계 (선택) · 숫자로 한 번 더 확인

6교시에서 배운 t-test 입니다.
**"할인 20% 초과 주문"과 "그 이하 주문"의 평균 이익이 정말 다른지** 봅니다.

```python
from scipy import stats

high_discount = clean[clean['Discount'] > 0.2]['Profit']
low_discount = clean[clean['Discount'] <= 0.2]['Profit']

t, p = stats.ttest_ind(high_discount, low_discount, equal_var=False)

print('할인 20% 초과 : {:>6}건   평균이익 {:8.1f}'.format(len(high_discount), high_discount.mean()))
print('할인 20% 이하 : {:>6}건   평균이익 {:8.1f}'.format(len(low_discount), low_discount.mean()))
print()
print('p 값 : {:.6f}'.format(p))
```

**p 값이 아주 작습니다. "우연히 이만큼 차이 날 리는 없다"는 뜻입니다.**

그런데 6교시에서 강조한 것을 다시 떠올려 주세요.
**p 값이 작다는 건 "차이가 있다"는 것이지 "차이가 크다"는 게 아닙니다.**
여기서는 차이의 크기 자체가 큽니다. 주문당 평균 이익이 **+48 에서 -96 으로** 뒤집힙니다.
p 값보다 이 숫자가 보고할 값입니다.

그리고 다시, **이건 인과가 아닙니다.** 할인이 이익을 깎았는지,
안 팔리는 물건에 할인을 붙인 것인지 이 데이터로는 구분할 수 없습니다.

엑셀이라면 파일을 복사해서 손으로 고칩니다. 고치다가 한 군데를 빠뜨립니다.

> ### 정리하면

### 문제 2. `Sales` 가 0 이하인 행을 제외하세요.

```python
step = step[step['Sales'] > 0]
print('남은 행:', len(step))
```

### 문제 3. `Sub-Category` 별 매출·이익 합계를 구하세요.

```python
by_sub = step.groupby('Sub-Category').agg(sales=('Sales', 'sum'), profit=('Profit', 'sum')).round(0)
by_sub.head()
```

### 문제 4. 이익률(이익 ÷ 매출 × 100) 열을 만들고 낮은 순으로 5개를 보세요.

```python
by_sub['margin'] = (by_sub['profit'] / by_sub['sales'] * 100).round(1)
by_sub.sort_values('margin').head(5)
```

### 문제 5. 적자 품목만 남기세요.

```python
by_sub[by_sub['margin'] < 0]
```

### 문제 6. 적자 품목의 이익률을 막대그래프로 그리고, 결론을 제목으로 다세요.

```python
loss = by_sub[by_sub['margin'] < 0]
loss['margin'].plot(kind='barh', figsize=(7, 3), color='#C44E52')
plt.title('세 품목이 팔수록 손해를 보고 있습니다')
plt.xlabel('이익률 (%)')
plt.show()
```
