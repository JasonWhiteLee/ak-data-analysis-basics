## 1) 변수 — 값에 이름을 붙인다

````python
sales = 1500
cost = 900

profit = sales - cost
profit
````

## 2) 값에는 종류가 있다 — 오늘 사고의 90%

- `1500` -> **숫자** (계산 가능)
- `'1500'` -> **글자** (따옴표가 붙으면 숫자가 아닙니다)

아래를 실행하고, 결과를 예상과 비교해 보세요.

````python
'1500' + '900'
````

`2400` 이 아니라 `'1500900'` 이 나옵니다.
파이썬은 글자끼리 더하라고 하면 **이어 붙입니다.**

`type()` 으로 종류를 확인할 수 있습니다.

````python
print(type(1500))
print(type('1500'))
print(type(1500.5))
````

> 엑셀에서 숫자가 왼쪽 정렬되고 `SUM` 이 0으로 나오던 그 상황과 **같은 문제**입니다.

# 1-4. 표를 담는 그릇 — DataFrame

pandas에서 표는 **DataFrame**입니다. **엑셀 시트 하나**라고 생각하면 정확합니다.

직접 하나 만들어 보겠습니다.

````python
df = pd.DataFrame({
    'branch': ['서울', '부산', '대구'],
    'sales': [1500, 900, 700],
    'staff': [12, 7, 5],
})

df
````

| 부르는 말 | 무엇 |
|---|---|
| **행 (row)** | 가로줄. 여기서는 지점 하나가 한 행 |
| **열 (column)** | 세로줄. `branch`, `sales`, `staff` |
| **인덱스** | 맨 왼쪽 `0 1 2`. 엑셀의 행 번호 (0부터) |

## 열 하나 꺼내기 — 대괄호

````python
df['sales']
````

이렇게 **열 하나**만 꺼낸 것을 **Series** 라고 부릅니다.
DataFrame이 엑셀 시트면, Series는 **한 열(column) 전체**입니다.

## 열 여러 개 꺼내기 — 대괄호 두 겹

리스트를 넣어야 하므로 `[[ ]]` 가 됩니다. 헷갈리는 지점이니 눈에 익혀 두세요.

````python
df[['branch', 'sales']]
````

## 열에 계산 시키기

Series에 점을 찍어 일을 시킵니다.

````python
print(df['sales'].sum())
print(df['sales'].mean())
print(df['sales'].max())
````

## 새 열 만들기

없는 열 이름에 값을 넣으면 **새로 생깁니다.**
엑셀에서 빈 열에 수식을 넣는 것과 같습니다.

````python
df['sales_per_staff'] = df['sales'] / df['staff']

df
````

> **줄 단위로 반복문을 쓰지 않았다는 점**을 보세요.
> pandas는 열 전체를 한 번에 계산합니다. 엑셀에서 수식을 아래로 드래그하는 것과 같습니다.
