## 2) 값에는 종류가 있다 — 오늘 사고의 90%

# 1-4. 표를 담는 그릇 — DataFrame

pandas에서 표는 **DataFrame**입니다. **엑셀 시트 하나**라고 생각하면 정확합니다.

직접 하나 만들어 보겠습니다.

````python
df = pd.DataFrame({
    'branch': ['서울', '부산', '대구'],
    'sales': [1500, 900, 700],
    'staff': [12, 7, 5],
})

df
````

| 부르는 말 | 무엇 |
|---|---|
| **행 (row)** | 가로줄. 여기서는 지점 하나가 한 행 |
| **열 (column)** | 세로줄. `branch`, `sales`, `staff` |
| **인덱스** | 맨 왼쪽 `0 1 2`. 엑셀의 행 번호 (0부터) |

## 열 하나 꺼내기 — 대괄호

````python
df['sales']
````

이렇게 **열 하나**만 꺼낸 것을 **Series** 라고 부릅니다.
DataFrame이 엑셀 시트면, Series는 **한 열(column) 전체**입니다.

## 열 여러 개 꺼내기 — 대괄호 두 겹

리스트를 넣어야 하므로 `[[ ]]` 가 됩니다. 헷갈리는 지점이니 눈에 익혀 두세요.

````python
df[['branch', 'sales']]
````

## 열에 계산 시키기

Series에 점을 찍어 일을 시킵니다.

````python
print(df['sales'].sum())
print(df['sales'].mean())
print(df['sales'].max())
````

## 새 열 만들기

없는 열 이름에 값을 넣으면 **새로 생깁니다.**
엑셀에서 빈 열에 수식을 넣는 것과 같습니다.

````python
df['sales_per_staff'] = df['sales'] / df['staff']

df
````

> **줄 단위로 반복문을 쓰지 않았다는 점**을 보세요.
> pandas는 열 전체를 한 번에 계산합니다. 엑셀에서 수식을 아래로 드래그하는 것과 같습니다.

---
## 여기서 아주 중요한 것 하나

이 표는 **한 줄이 주문 하나가 아닙니다.**
한 번 주문에 여러 물건을 담으면, **물건마다 한 줄씩** 생깁니다.

직접 확인해 보세요.

````python
print('전체 행 수 :', len(orders))
print('주문 건수  :', orders['Order ID'].nunique())
````

숫자가 다릅니다. **행 수 != 주문 수** 입니다.

이걸 모르면 이런 보고가 나갑니다.

> *"올해 주문이 10,239건 발생했습니다"* -> **틀렸습니다.** 그건 품목 줄 수입니다.

> ### 오늘의 첫 번째 교훈
> **표의 한 줄이 무엇을 뜻하는지 먼저 확인합니다.**
> 주문인지, 품목인지, 고객인지, 하루인지. 이걸 안 물어보고 센 숫자는 대개 틀립니다.

한 주문이 여러 줄인 예를 직접 보겠습니다.

# 1-8. 에러는 사고가 아닙니다

오늘 하루 에러를 수십 번 만납니다. 정상입니다.
**에러 메시지는 혼내는 말이 아니라, 어디가 막혔는지 알려 주는 안내문입니다.**

일부러 세 가지를 내 보겠습니다. 실행하고 **맨 마지막 줄만** 읽으세요.

### 1) 없는 열을 불렀을 때

````python
orders['Revenue']
````

```
KeyError: '매출액'
```
**"'매출액' 이라는 열이 없다"** 는 뜻입니다. 이 표의 매출 열 이름은 `Sales` 였죠.
`orders.columns` 로 이름을 다시 확인하면 됩니다.

### 2) 만들지 않은 이름을 불렀을 때

````python
sales_total
````

```
NameError: name 'sales_total' is not defined
```
**"그런 이름을 만든 적이 없다"** 는 뜻입니다.
-> 대개 **위쪽 셀을 실행하지 않고 건너뛴 경우**입니다. **런타임 > 모두 실행**으로 해결됩니다.

### 3) 오타가 났을 때

````python
orders.hed()
````

```
AttributeError: 'DataFrame' object has no attribute 'hed'
```
**"DataFrame에는 `hed` 라는 기능이 없다"** — `head` 오타입니다.

> ## 에러를 읽는 요령
> 1. 빨간 글씨가 길어도 **맨 아랫줄**부터 읽습니다
> 2. 에러 **이름**(`KeyError`, `NameError`, `AttributeError`)이 종류를 알려 줍니다
> 3. 모르겠으면 **그 줄을 그대로 복사**해서 검색하거나 AI에게 붙여 넣습니다
